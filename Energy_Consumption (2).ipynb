```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Project libraries imported successfully!")
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00374/energydata_complete.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nFirst five rows:")
display(df.head())
```

```python
print("Dataset information:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nStatistical summary:")
display(df.describe())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Appliances"], bins=30)

plt.title("Distribution of Appliance Energy Consumption")
plt.xlabel("Appliance energy use (Wh per 10 minutes)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
print("Dataset columns:")
print(df.columns.tolist())

print("\nTarget variable:")
print("Appliances")

print("\nCorrelation with appliance energy use:")
correlations = df.corr(numeric_only=True)["Appliances"].drop("Appliances")
correlations = correlations.sort_values(key=abs, ascending=False)

display(correlations.head(10).to_frame("Correlation"))
```

```python
plt.figure(figsize=(10, 6))

correlations.sort_values().plot(kind="barh")

plt.title("Correlation of Variables with Appliance Energy Use")
plt.xlabel("Correlation coefficient")
plt.ylabel("Dataset variables")
plt.axvline(x=0, linestyle="--")

plt.tight_layout()
plt.show()
```

```python
# Separate the target from the input features
y = df["Appliances"]

# Exclude the target and raw date column from the features
X = df.drop(columns=["Appliances", "date"])

print("Feature data shape:", X.shape)
print("Target data shape:", y.shape)

print("\nFirst five feature rows:")
display(X.head())

print("\nFirst five target values:")
display(y.head())
```

```python
from sklearn.model_selection import train_test_split

# Keep observations in chronological order
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training targets:", y_train.shape)
print("Testing targets:", y_test.shape)

print("\nTraining data ends at row:", split_index - 1)
print("Testing data starts at row:", split_index)
```

```python
from sklearn.linear_model import LinearRegression

# Create the model
model = LinearRegression()

# Train it using the training data only
model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
print("Number of training rows:", len(X_train))
print("Number of input features:", X_train.shape[1])
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Predict on the unseen test data
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Linear Regression Model Evaluation")
print("----------------------------------")
print(f"MAE:  {mae:.3f} Wh per 10 minutes")
print(f"RMSE: {rmse:.3f} Wh per 10 minutes")
print(f"R²:   {r2:.3f}")
```

```python
plt.figure(figsize=(12, 5))

actual_values = y_test.iloc[:100].to_numpy()
predicted_values = np.asarray(y_pred[:100])

plt.plot(actual_values, label="Actual energy use")
plt.plot(predicted_values, label="Predicted energy use")

plt.title("Actual vs Predicted Appliance Energy Use")
plt.xlabel("Test observation (10-minute interval)")
plt.ylabel("Appliance energy use (Wh)")
plt.legend()
plt.tight_layout()
plt.show()
```

Linear Regression Model Evaluation

The Linear Regression model achieved an MAE of 52.015 Wh per measurement interval and an RMSE of 86.730 Wh per measurement interval. The higher RMSE suggests that some prediction errors may be considerably larger than others.

The R² score was 0.092, meaning the model explains approximately 9.2% of the variation in appliance energy use relative to a test-set mean baseline. This indicates limited predictive performance and suggests that further model improvement is needed.

Future work will explore additional time-based features and alternative regression algorithms. Any improved model will be evaluated using the same chronological test period for a fair comparison.

Final Project Conclusion

This project applied machine learning techniques to predict household appliance energy consumption using the UCI Appliances Energy Prediction dataset. Two regression models, Linear Regression and Decision Tree Regression, were trained and evaluated using Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and the coefficient of determination (R²).

The Linear Regression model achieved an MAE of 52.015 Wh per 10 minutes, an RMSE of 86.730 Wh per 10 minutes, and an R² of 0.092. The Decision Tree model performed worse on the test data, with an MAE of 155.845 Wh per 10 minutes, an RMSE of 211.575 Wh per 10 minutes, and an R² of -4.401.

Based on these results, Linear Regression performed better than the Decision Tree model on this test set. However, its low R² indicates that further improvement is needed. Future work could explore additional features, alternative machine learning algorithms, and hyperparameter tuning to improve prediction accuracy.
